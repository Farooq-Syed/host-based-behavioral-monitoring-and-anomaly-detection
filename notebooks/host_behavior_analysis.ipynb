# Host Behavior Analysis

This notebook explores the sample host telemetry dataset used for host-based behavioral monitoring and anomaly detection.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/sample_host_telemetry.csv')
df.head()
```

```python
df['label'].value_counts()
```

```python
features = ['cpu_percent', 'disk_write_mb', 'file_rename_count', 'suspicious_extension_changes', 'entropy_score']
df.groupby('label')[features].mean().T
```

```python
colors = df['label'].map({'benign': '#1d3557', 'suspicious': '#c1121f'})
plt.figure(figsize=(8, 5))
plt.scatter(df['disk_write_mb'], df['file_rename_count'], c=colors)
plt.xlabel('disk_write_mb')
plt.ylabel('file_rename_count')
plt.title('Disk Writes vs File Renames')
plt.show()
```
